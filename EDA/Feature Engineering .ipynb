```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\rampa\DS Python Class\Baking Loan\Data\Cleande data\banking_loan_cleand_data.csv")
```

### Age Group

Age groups are created to categorize customers into meaningful age ranges.
This makes it easier to analyze customer behavior, loan patterns, approval rates, and payment delays across different age groups.


```python
def age_group(age):
    if age <= 25:
        return "Young"
    elif age <= 40:
        return "Adult"
    elif age <= 55:
        return "Senior"
    else:
        return "Older"

df["Age_Group"] = df["age"].apply(age_group)
df["Age_Group"].unique()
```

### Application Year

The `Application_Year` column is created from the `Application_Date` column to analyze loan applications on a year-wise basis. This helps identify trends and changes in loan applications over different years.


```python
#Convert Application Date

#When data is imported from a CSV file, the Application_Date column may not be recognized as a date datatype.
#Therefore, it is converted to the datetime datatype before performing date-based analysis such as extracting the application year.

df["application_date"] = pd.to_datetime(df["application_date"])
df["application_Year"] = df["application_date"].dt.year

# Check the result
df[["application_date", "application_Year"]].head()
```

### Application Month

The `Application_Month` column is created from the `Application_Date` column to analyze loan applications on a month-wise basis. 
This helps identify monthly trends and patterns in loan applications.


```python
df["Application_Month"] = df["application_date"].dt.month_name()
df[["application_date", "Application_Month"]].head()
```

### Income Segment

The `Income_Segment` column is created by grouping customers based on their annual income. This makes it easier to analyze loan applications, approval status, loan amounts, and payment behavior across different income groups.


```python
def income_segment(income):
    if income <= 300000:
        return "Low"
    elif income <= 600000:
        return "Medium"
    else:
        return "High"

df["Income_Segment"] = df["annual_income"].apply(income_segment)
```

### Credit Score Segment

The `Credit_Score_Segment` column is created to categorize customers based on their credit scores. This helps analyze loan approval, loan status, and payment behavior across different credit score categories.


```python
def credit_score_band(score):
    if score < 580:
        return "Poor"
    elif score < 670:
        return "Fair"
    else:
        return "Good"

df["Credit_Score_Band"] = df["credit_score"].apply(credit_score_band)
```

### Payment Delay Category

The `Delay_Category` column is created by grouping customers based on the number of payment delay days. This helps analyze customer payment behavior and identify customers with different levels of payment delays.


```python
def delay_category(days):
    if days == 0:
        return "No Delay"
    elif days <= 30:
        return "Low Delay"
    elif days <= 60:
        return "Moderate Delay"
    else:
        return "High Delay"

df["Delay_Category"] = df["payment_delay_days"].apply(delay_category)
```
