```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\rampa\DS Python Class\Baking Loan\Data\Row Data\banking_loan_row_data.csv")
```

## Standarize Columns Name 

```python
df.rename(columns={
    "Customer_ID": "customer_id",
    "Application_Date": "application_date",
    "Age": "age",
    "Gender": "gender",
    "Employment_Type": "employment_type",
    "Annual_Income": "annual_income",
    "City": "city",
    "Credit_Score": "credit_score",
    "Loan_Type": "loan_type",
    "Loan_Tenure_Years": "loan_tenure_years",
    "Loan_Amount": "loan_amount",
    "Approval_Status": "approval_status",
    "Loan_Status": "loan_status",
    "Payment_Delay_Days": "payment_delay_days"
}, inplace=True)

# Check column names
df.columns
```


### Missing Value Analysis
The dataset does not contain any missing values. All columns have complete data.


### Duplicate Row Analysis

The dataset does not contain any duplicate rows. All records are unique.


## Data Messiness

```python
print(df["city"].unique().tolist())
print("=====================================================================================================================")
df["city"] = (
    df["city"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)
print(df["city"].unique().tolist())
```

```python
print(df["employment_type"].unique().tolist())
print("=====================================================================================================================")
df["employment_type"] = (
    df["employment_type"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)
print(df["employment_type"].unique().tolist())
```

```python
print("=====================================================================================================================")
df["loan_amount"] = (
    df["loan_amount"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.strip()
)
```

## Date type Coverson 

```python
df["loan_amount"] = pd.to_numeric(df["loan_amount"], errors="coerce")
df["loan_amount"].dtypes

df["application_date"] = pd.to_datetime(df["application_date"])
df["application_date"].dtypes
```
